## Problem Statement

Currently, city maintenance is largely reactive rather than preventative. Using the 2015 NYC Street Tree Census dataset, this project builds a machine learning model to predict tree health status (Good, Fair, Poor) and analyze its impact on surrounding sidewalk infrastructure. The goal is to identify key physical and environmental risk factors, enabling urban planners to prioritize proactive maintenance and reduce structural hazards.

### Class Imbalance Handling Strategy

The target variable `health` shows extreme class imbalance, heavily dominated by the `Good` class (~80%), with significantly fewer `Fair` (~15%) and `Poor` (~5%) records[cite: 3]. Standard models trained on this raw data will yield artificially high overall accuracy while failing to identify unhealthy trees.

* **Stratified Data Splitting:** Perform train-test splits using stratification (`stratify=y`) to maintain the baseline class distribution across training and testing sets.
* **Model-Level Cost-Sensitive Learning:** Apply class weights (`class_weight='balanced'`) in tree algorithms (e.g., Random Forest, XGBoost) to heavily penalize misclassifications of minority classes (`Poor` and `Fair`).
* **Synthetic Oversampling (SMOTE):** Use `SMOTE` (or `SMOTENC` for categorical features) exclusively on the training set to synthetically increase the representation of `Poor` and `Fair` classes before model training.
* **Evaluation Metric Shift:** Ignore standard Accuracy and optimize models using **Macro F1-Score**, **Balanced Accuracy**, and **Per-Class Recall** via a Confusion Matrix.

### Missing Data Strategy & Handling

* **Target & Critical Attributes (`health`, `sidewalk`, `spc_common`, `spc_latin`):**
  * **Missing Count:** ~31,616 to 31,619 rows (~4.62%)
  * **Reason & Action:** These values are missing mainly because the entry represents a stump or dead tree. Since `health` and `sidewalk` are core target variables for prediction, rows with missing values in these columns were **dropped** (`dropna(subset=['health', 'sidewalk', 'spc_common'])`).

* **Optional Categorical Attributes (`steward`, `guards`):**
  * **Missing Count:** `steward` (~75.96%), `guards` (~88.32%)
  * **Reason & Action:** High null counts represent trees where no specific guard or steward program was present. These missing entries were **imputed** with the explicit label `'None'` (`fillna('None')`) to retain the records without introducing synthetic bias.

* **Problem & Environmental Flags (`root_*`, `trnk_*`, `brch_*`):**
  * **Action:** Unrecorded values in binary problem flags were converted to `False` (where explicit `'Yes'` values became `True`), treating absent notes as no observed damage.

```python
# Which features are most correlated with your target, and with each other?
# Map ordinal target to numeric for correlation calculation
df_corr = df.copy()
health_map = {'Poor': 0, 'Fair': 1, 'Good': 2}
df_corr['health_num'] = df_corr['health'].map(health_map)

# Convert boolean columns to int
for col in bool_features:
    df_corr[col] = df_corr[col].astype(int)

# Compute correlations with target
numeric_cols = df_corr.select_dtypes(include=['int64', 'float64', 'int32']).columns
target_corr = df_corr[numeric_cols].corr()['health_num'].sort_values(ascending=False)

print("--- Correlation with Health Target ---")
print(target_corr.drop('health_num'))

# Plot full correlation heatmap for numerical/boolean variables
plt.figure(figsize=(12, 8))
sns.heatmap(df_corr[numeric_cols].corr(), annot=False, cmap='coolwarm', vmin=-1, vmax=1)
plt.title('Feature Correlation Matrix')
plt.show()
```

### Correlation Analysis Takeaways

* **Strongest Physical Indicator:** `tree_dbh` shows the highest meaningful correlation with `health_num` (~0.074), reflecting that tree maturity correlates slightly with overall health status.
* **Data Leakage Artifact:** `tree_id` exhibits a ~0.115 correlation purely due to sequential data collection ordering; it carries no real domain signal and will be dropped before modeling.
* **Low Linear Signal in Spatial IDs:** Administrative and location variables (`borocode`, `community board`, `latitude`) show minimal linear relationship ($\le 0.03$) with health, highlighting the need for non-linear models (like Tree-Based models) or feature engineering.
* **Zero Variance Feature:** `stump_diam` resulted in `NaN` correlation because stump rows were filtered out during data cleaning, leaving no variance in the column.

```python
# Are there outliers? how would you handle them?
# Summary statistics and extreme quantile inspection for tree diameter
print("--- Tree Diameter Summary Statistics ---")
print(df['tree_dbh'].describe(percentiles=[0.5, 0.95, 0.99, 0.999]))

# Visualize outliers using Boxplot and Histogram
fig, axes = plt.subplots(1, 2, figsize=(12, 4))
sns.boxplot(x=df['tree_dbh'], ax=axes[0], color='skyblue')
axes[0].set_title('Boxplot of tree_dbh')

sns.histplot(df['tree_dbh'], bins=50, ax=axes[1], kde=True)
axes[1].set_title('Distribution of tree_dbh')
plt.show()
```

### Outlier Analysis Takeaways (`tree_dbh`)

### Outlier Handling Strategy (`tree_dbh`)

* **Trimming Implausible Values:** Drop measurement errors where `tree_dbh > 50` inches (e.g., maximum of 450 inches) to eliminate biologically impossible trunk sizes.
* **Zero-Value Handling:** Filter out or impute `tree_dbh == 0` entries, which represent unrecorded saplings or missing measurements.
* **Log Transformation:** Apply a $\log(x + 1)$ transformation to contract the right-skewed distribution for linear models.
